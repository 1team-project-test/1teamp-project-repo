```python
import json
import urllib3
import pandas as pd
import requests
import xmltodict
```

```python
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
```

```python
url = "http://apis.data.go.kr/B552657/ErmctInfoInqireService/getEmrrmRltmUsefulSckbdInfoInqire"
```

```python
apikey = "C:/GG/Python/project/api.txt"
with open(apikey, "r", encoding="utf-8") as f:
    apikey = f.read().strip()
```

```python
# 4. 요청 파라미터 설정
params = {
    "serviceKey": apikey,  # 파일에서 읽어온 실제 인증키가 들어갑니다.
    "STAGE1": "서울특별시",
    "STAGE2": 0,
    "pageNo": "1",
    "numOfRows": "10",
}
```

```python
try:
    # 5. API 호출
    response = requests.get(url, params=params, verify=False)

    if response.status_code == 200:
        # XML 데이터를 파이썬 딕셔너리로 변환
        dict_data = xmltodict.parse(response.text)

        # 6. 데이터 계층 접근 (response -> body -> items -> item)
        try:
            items_wrapper = dict_data["response"]["body"]["items"]

            if items_wrapper is None:
                print("검색 결과가 없습니다. 지역명을 다시 확인해 보세요.")
            else:
                items = items_wrapper["item"]

                # 결과가 1개일 때를 대비해 리스트로 감싸기
                if isinstance(items, dict):
                    items = [items]

                # 7. 판다스 데이터프레임 변환 및 출력
                df = pd.DataFrame(items)

                print(
                    f"--- 데이터 가져오기 성공! 총 {len(df)}개의 기관 정보 수집됨 ---"
                )
                display(df.head())

        except KeyError:
            # 인증키가 시스템에 등록 완료되지 않았거나 이중 인코딩 문제가 있을 때
            print("--- 서버가 에러 메시지를 반환했습니다 ---")
            print(json.dumps(dict_data, indent=2, ensure_ascii=False))

    else:
        print(f"호출 실패 (Status Code): {response.status_code}")

except Exception as e:
    print(f"코드 실행 중 에러 발생: {e}")
```
